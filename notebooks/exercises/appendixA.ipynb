## Exercise A.3
How many parameters does the neural network introduced in the following listing have ?

```python
import torch.nn as nn

class NeuralNetwork(nn.Module):
    def __init__(self, n_inputs, n_outputs):
        super().__init__()
        self.layers = nn.Sequential(
            nn.Linear(n_inputs, 30),
            nn.ReLU(),
            nn.Linear(30, 20),
            nn.ReLU(),
            nn.Linear(20, n_outputs),
        )

    def forward(self, x):
        logits = self.layers(x)
        return logits
```

```python
net = NeuralNetwork(2, 2)
print(sum(param.numel() for param in net.parameters()))
```

## Exercise A.4
Compare the run time of matrix multiplication on a CPU to a GPU. At what matrix size do you begin to see the matrix multiplication on the GPU being faster than on the CPU? Hint: use the `%timeit` command in Jupyter to compare the run time. For example, given matrices a and b, run the command `%timeit a @ b` in a new notebook cell.

```python
import torch

torch.manual_seed(123)

size = 75
a = torch.randn(size, size)
b = torch.randn(size, size)

a_gpu = a.detach().clone().to('cuda')
b_gpu = b.detach().clone().to('cuda')
```

```python
%timeit a @ b
```

```python
%timeit a_gpu @ b_gpu
```
