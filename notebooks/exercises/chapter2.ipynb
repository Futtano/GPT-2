## Exercise 2.1
Try the BPE tokenizer from the tiktoken library on the unknown words “Akwirw ier” and print the individual token IDs. Then, call the decode function on each of the resulting integers in this list. Lastly, call the decode method on the token IDs to check whether it can reconstruct the original input, “Akwirw ier.”

```python
import tiktoken

tokenizer = tiktoken.get_encoding('gpt2')
token_ids = tokenizer.encode('Akwirw ier')
print("Token IDs: ", token_ids)
tokens = [tokenizer.decode((token_id,)) for token_id in token_ids]
print("Tokens: ", tokens)
text = tokenizer.decode(token_ids)
print("Reconstructed text: ", text)
```

### Exercise 2.2
To develop more intuition for how the data loader works, try to run it with different settings such as `max_length=2` and `stride=2`, and `max_length=8` and `stride=2`.

```python
from gpt_2.dataset import (
    create_data_loader_v1,
    download_pt_dataset
)

# Get the text dataset
file_path = download_pt_dataset()
with open(file_path, 'r', encoding='utf-8') as fp:
    txt = fp.read()

# Dataloader with max_length=2 and stride=2
dataloaderv1 = create_data_loader_v1(
    txt,
    batch_size=4,
    max_length=2,
    shuffle=False,
    stride=2)
it = iter(dataloaderv1)
batchv1 = next(it)
print(f"DataLoader(max_length=2, stride=2):\n", batchv1)

# Dataloader with max_length=8 and stride=2
dataloaderv2 = create_data_loader_v1(
    txt,
    batch_size=4,
    max_length=8,
    shuffle=False,
    stride=2)
it = iter(dataloaderv2)
batchv2 = next(it)
print(f"DataLoader(max_length=8, stride=2):\n", batchv2)
```
